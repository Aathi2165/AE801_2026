# Orbital Mechanics - Keplerian orbit

#### This is sample code to automate manual Keplerian 2D orbit calculation.

```python
import numpy as np
import matplotlib.pyplot as mp
```

#### Initial condition

$$
\mu = 398600.4418\ {\rm km^3/s^2}
$$

$$
\mathbf{r}_0 =
\begin{bmatrix}
7000\\
0\\
0
\end{bmatrix}
\ {\rm km}
$$

$$
\mathbf{v}_0 =
\begin{bmatrix}
0\\
7.5\\
0
\end{bmatrix}
\ {\rm km/s}
$$


```python
mu = 398600.4418       # km^3/s^2

r0 = np.array([7000.0, 0.0, 0.0])    # km
v0 = np.array([0.0, 7.5, 0.0])       # km/s
r = np.linalg.norm(r0)
v = np.linalg.norm(v0)
```

```python
H_v = np.cross(r0, v0)
H = np.linalg.norm(H_v)
```

```python
print("H_v is ",H_v)
print("H is ", H,"Km^2/s")
```

```python
e_v = (np.cross(v0, H_v)/mu)-(r0/r)
e = np.linalg.norm(e_v)
print("Eccentricity vector is ",e_v)
print("Ecentricity is ",e)
```
